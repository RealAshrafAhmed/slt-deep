# Phase Transition Analysis

*Example notebook demonstrating clean architecture - imports only from libraries.*

```python
# ✅ GOOD: Import from libraries only
import numpy as np
from kl import kl_divergence_gaussian
from scipy_extensions import normal_mixture

# ❌ BAD: Never do this!
# from ..quasi_singular_models.some_module import function
# ↑ Experiments should never import from each other
```

```python
# Example analysis using shared libraries
np.random.seed(42)
data = np.random.normal(0, 1, 1000)

# Use global library
mixture = normal_mixture.fit_em(data, n_components=2)
print(f"Fitted mixture: {mixture}")

# Use domain-specific library
kl_div = kl_divergence_gaussian(mu1=0, sigma1=1, mu2=0.1, sigma2=1.2)
print(f"KL divergence: {kl_div:.4f}")
```

## Clean Architecture Benefits

1. **No tight coupling** - experiments can't accidentally depend on each other
2. **Forced abstractions** - shared code must be properly packaged in libraries
3. **Clear dependencies** - easy to see what libraries an experiment uses
4. **Better testing** - libraries can have comprehensive tests
5. **Reusable code** - libraries can be used by many experiments
